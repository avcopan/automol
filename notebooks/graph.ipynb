```python
smiles = "C=C"
threshold = 4e-3
method1 = "xtb"
method2 = "xtb"
```

```python
from automol import Geometry, geom, Identity
from qccompute import compute
from qcdata import ProgramInput, CalcType, Model, Structure

from IPython.display import display

BOHR_TO_ANGSTROM = 0.5291772109

geo1 = Identity(kind="stereoisomer", algorithm="rdkit smiles", value=smiles).geometry()
ident1 = ProgramInput(structure=
    Structure(symbols=geo1.symbols, geometry=geo1.coordinates.astype(list) / BOHR_TO_ANGSTROM),
    calctype=CalcType.optimization,
    model=Model(method=method1)
)
out1 = compute("orca", ident1)
struc1 = out1.data.final_structure
geo1 = Geometry(symbols=struc1.symbols, coordinates=struc1.geometry * BOHR_TO_ANGSTROM, charge=0, spin=0)

geo2 = Identity(kind="stereoisomer", algorithm="rdkit smiles", value=smiles).geometry()
ident2 = ProgramInput(structure=
    Structure(symbols=geo2.symbols, geometry=geo2.coordinates.astype(list) / BOHR_TO_ANGSTROM),
    calctype=CalcType.optimization,
    model=Model(method=method2)
)
out2 = compute("orca", ident2)
struc2 = out2.data.final_structure
geo2 = Geometry(symbols=struc2.symbols, coordinates=struc2.geometry * BOHR_TO_ANGSTROM, charge=0, spin=0)

print(geom.transform.is_equivalent(geo1, geo2, threshold=threshold))
display(geom.view(geo1, label=True))
display(geom.view(geo2, label=True))
```

```python
from stereomolgraph import StereoMolGraph
from stereomolgraph.coords import Geometry as SmgGeometry
from stereomolgraph.algorithms.isomorphism import vf2pp_all_isomorphisms

geo = SmgGeometry(atom_types = geo1.atomic_numbers, coords = geo1.coordinates)
smg = StereoMolGraph.from_geometry(geo)

for iso in vf2pp_all_isomorphisms(smg, smg, stereo=True):
    print(iso)
```
