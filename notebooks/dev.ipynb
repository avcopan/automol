```python
from automol import Identity
from stereomolgraph import StereoMolGraph
from stereomolgraph.coords import Geometry

ident = Identity(kind="stereoisomer", algorithm="rdkit smiles", value="[CH2]CCC1OC1")
geo = ident.geometry()

geo_smg = Geometry(atom_types=geo.atomic_numbers, coords=geo.coordinates)

gra = StereoMolGraph.from_geometry(geo=geo_smg)

gra
```
